```python
import numpy as np
import rebound
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from astropy.timeseries import LombScargle

# --- SETTINGS ---
NUM_SIMS = 1000
MAX_TRANSITS = 100
DT = 0.01

def get_ttv_data():
    print("Generating data...")
    all_x, all_y = [], []
    for i in range(NUM_SIMS):
        try:
            # Physics: Planet B and a hidden Planet C near 2:1 resonance
            pb = np.random.uniform(1.0, 3.0)
            ratio = 2.0 + np.random.uniform(-0.01, 0.01) # Near resonance
            mc = np.random.uniform(1e-4, 5e-3) # Mass range
            
            sim = rebound.Simulation()
            sim.dt = DT
            sim.add(m=1.0)
            sim.add(m=1e-4, P=pb)
            sim.add(m=mc, P=pb * ratio)
            sim.move_to_com()

            # Simple Transit Detection
            transits = []
            while len(transits) < MAX_TRANSITS:
                sim.integrate(sim.t + pb * 0.9) # Jump close to next transit
                while sim.particles[1].y < 0: sim.integrate(sim.t + DT) # Crossing
                transits.append(sim.t)

            # Calculate TTV and Periodogram
            ttv = np.array(transits) - (np.arange(MAX_TRANSITS) * pb + transits[0])
            freqs = np.linspace(1/50, 1/3, 100)
            power = LombScargle(np.arange(MAX_TRANSITS), ttv).power(freqs)

            # Features: [Total Wobble Strength, Frequency Signature]
            all_x.append(np.hstack(([np.std(ttv)], power)))
            all_y.append(mc)
        except: continue
    return np.array(all_x), np.array(all_y)

# --- THE SIMPLIFIED MODEL ---
class CompactPredictor(nn.Module):
    def __init__(self, in_size):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_size, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )
    def forward(self, x): return self.net(x)

# --- RUN ---
X_raw, y_raw = get_ttv_data()
y_log = np.log10(y_raw * 332946.0) # Target: Log Earth Masses
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_raw)

xtrain, xtest, ytrain, ytest = train_test_split(X_scaled, y_log, test_size=0.2)

model = CompactPredictor(xtrain.shape[1])
opt = optim.Adam(model.parameters(), lr=0.001)

for e in range(101):
    model.train()
    p = model(torch.Tensor(xtrain)).flatten()
    loss = nn.MSELoss()(p, torch.Tensor(ytrain))
    opt.zero_grad(); loss.backward(); opt.step()
    if e % 20 == 0: print(f"Epoch {e}, Loss: {loss.item():.4f}")

# Plot
model.eval()
preds = model(torch.Tensor(xtest)).detach().numpy().flatten()
plt.scatter(ytest, preds, alpha=0.5)
plt.plot([ytest.min(), ytest.max()], [ytest.min(), ytest.max()], 'r')
plt.title("Hallelujah Version: Simplified Mass Prediction")
plt.xlabel("Actual (Log Earth Mass)"); plt.ylabel("Predicted")
plt.show()
```
